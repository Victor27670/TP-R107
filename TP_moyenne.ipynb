```python
moy_1=0
som_1=0
coe_1=0

for i in range(101,104):
    note=float(input("Donnez-moi la note de la ressource R{}".format(i)))
    coeff=float(input("Donnez-moi le coefficient de la ressource R{}".format(i)))
    som_1+=note*coeff
    coe_1+=coeff
for i in range(11,13):
    note=float(input("Donnez-moi la note de la SAE{}".format(i)))
    coeff=float(input("Donnez-moi le coefficient de la SAE{}".format(i)))
    som_1+=note*coeff
    coe_1+=coeff
moy_1=som_1/coe_1
print(moy_1)
```
